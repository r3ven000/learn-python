```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    'Тематика': ['Веб-сайты', 'Телеграм-боты', 'Анализ данных', 'Игры на PyGame'],
    'Количество': [15, 22, 12, 8]
}

df = pd.DataFrame(data)

plt.figure(figsize=(6, 6))
plt.pie(
    df['Количество'],
    labels=df['Тематика'],
    autopct='%1.1f%%',
    startangle=140
)

plt.title('распределение проектов')

plt.show()
```
